```python
import pandas as pd
from sqlalchemy import create_engine,text
import matplotlib.pyplot as plt
import seaborn as sns
import streamlit as st
```

```python
df['Date'] = df['Date'].dt.date
```

```python
df.drop(columns = ['Vehicle Images'], inplace = True)
```

```python
df.describe()
```

```python
100*df.isnull().sum()/df.shape[0]
```

```python
df.info()
```

```python
num_columns = ['Driver_Ratings','Customer_Rating','V_TAT','C_TAT']
df[num_columns] = df[num_columns].fillna(0)

str_columns = ['Canceled_Rides_by_Customer','Canceled_Rides_by_Driver','Incomplete_Rides','Incomplete_Rides_Reason','Payment_Method']
df[str_columns] = df[str_columns].fillna('Unavailable')
```

```python
100*df.isnull().sum()/df.shape[0]
```

```python
df.value_counts("Payment_Method")
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Time'] = pd.to_datetime(df['Time'],format = '%H:%M:%S')
```

```python
df.info()
```

```python
from sqlalchemy import create_engine,text
base_engine = create_engine('mysql+pymysql://root:shoulderblade@localhost:3306/mysql')

with base_engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS ola_ride_insights"))
    conn.commit()

engine = create_engine('mysql+pymysql://root:shoulderblade@localhost:3306/ola_ride_insights')

df.to_sql('ola_ride_insights', con=engine, if_exists = 'replace', index=False)
```
